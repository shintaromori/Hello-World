## 💱 **Step 3：S&P500 の為替ヘッジを考える**

S&P500 はドル建ての資産です。
日本の投資家が S&P500 に投資すると、その価値は

* **株価変動（S&P500 の値動き）**
* **為替変動（USD/JPY の変動）**

の両方の影響を受けます。

ここでは、次の 2 通りの投資方法を比較します。

---

### ●（A）為替ヘッジなし（円換算）の S&P500

S&P500 にドル建てで投資し、為替リスクも抱えたままのケースです。
円換算した投資価値は

$$
(1 + R^{yen}_t) = (1 + R^{$}_t)(1 + R^{FX}_t)
$$

したがって：

$$
R^{yen}_t = (1 + R^{$}_t)(1 + R^{FX}_t) - 1
$$

> ※ よく使われる近似式
> $$
> R^{yen}_t \approx R^{$}_t + R^{FX}_t
> $$
> も参考として示しておきます。

---

### ●（B）為替ヘッジあり（1ドル買い、1ドル売りポート）

S&P500 をドル建てで買うと同時に、
**同額のドルを先物（またはフォワード）で売る**ことで為替変動を相殺します。

この場合、円ベースの投資価値は為替変動を受けず、
**純粋に S&P500 の値動きだけが反映されます。**

$$
R^{hedged}_t = R^{$}_t
$$

## 🔢 **年次リターンの計算**

株価（年次データ）は非常にシンプルです：

$$
R_{t} = \frac{P_t}{P_{t-1}} - 1
$$

為替（USDJPY）は月次データですが、
**“年次リターン = 年末値の変化率”** として計算します。

---

```python
# ============================
# Step 2：リターンとリスクの計算
# ============================

# --- 年次株価リターン（Nikkei & S&P500） ---
df_returns = pd.DataFrame(index=df.index)
df_returns["Nikkei Return"] = df["Nikkei"].pct_change()
df_returns["SP500 Return"]  = df["S&P500"].pct_change()

# --- 年次為替リターン（USDJPY） ---
# 年末レートを抽出（df2 は月次 → 年次にリサンプル）
df_fx_annual = df2.resample("Y").last()
df_fx_annual.index = df_fx_annual.index.year  # 2024-12-31 → 2024

df_returns["FX Return"] = df_fx_annual["Exchange Rate"].pct_change()

# --- 代表表示 ---
```

  → 為替と株価は独立して動くことが多く、

  に重要な役割を果たす。

# ✅ **リターン分布 + 正規分布フィットのプロット（完成版）**

```python
from scipy.stats import norm

def plot_return_distribution(returns, asset_name="Asset", bins=25):
    """
    returns : リターン（Series）
    asset_name : 資産名（表示用）
    bins : ヒストグラムのビン数
    """

    # NaN を除去
    data = returns.dropna()

    # 平均と標準偏差（年間ボラ）
    mu = data.mean()
    sigma = data.std()

    # 正規分布フィット用の x 範囲
    x = np.linspace(data.min(), data.max(), 400)
    pdf = norm.pdf(x, mu, sigma)

    # ---- 描画 ----
    plt.figure(figsize=(6,4))

    # ヒストグラム（確率密度として表示）
    plt.hist(data, bins=bins, density=True, alpha=0.6, color="skyblue", label="Histogram")

    # 正規分布フィット
    plt.plot(x, pdf, "r--", linewidth=2, label=f"N({mu:.3f}, {sigma:.3f}²)")

    plt.title(f"{asset_name}：リターン分布と正規分布フィット", fontsize=14)
    plt.xlabel("Return", fontsize=12)
    plt.ylabel("Density", fontsize=12)
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()

    print(f"平均リターン: {mu:.4f}")
    print(f"標準偏差（ボラ）: {sigma:.4f}")
```

# 📌 **解説（学生向けに言語化すると…）**

* ヒストグラムは「過去のリターンの実際のばらつき」を示す
* 赤い破線は「平均と標準偏差から作った理論的な正規分布」
* 実データと正規分布を比較すると、

  * ファットテール（裾が厚い）
  * 歪み（左右非対称）
    などが確認できる
* 金融リターンが「正規分布では表現しきれない」ことを直感的に理解できる

---


````python
from scipy.stats import norm

def plot_return_distribution(returns, asset_name="Asset", bins=25):
    """
    returns : リターン（Series）
    asset_name : 資産名（表示用）
    bins : ヒストグラムのビン数
    """

    # NaN を除去
    data = returns.dropna()

    # 平均と標準偏差（年間ボラ）
    mu = data.mean()
    sigma = data.std()

    # 正規分布フィット用の x 範囲
    x = np.linspace(data.min()-0.1, data.max()+0.1, 400)
    pdf = norm.pdf(x, mu, sigma)

    # ---- 描画 ----
    plt.figure(figsize=(6,4))

    # ヒストグラム（確率密度として表示）
    plt.hist(data, bins=bins, density=True, alpha=0.6, color="skyblue", label="Histogram")

    # 正規分布フィット
    plt.plot(x, pdf, "r--", linewidth=2, label=f"N({mu:.3f}, {sigma:.3f}²)")

    plt.title(f"{asset_name}：リターン分布と正規分布フィット", fontsize=14)
    plt.xlabel("Return", fontsize=12)
    plt.ylabel("Density", fontsize=12)
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()

    print(f"平均リターン: {mu:.4f}")
    print(f"標準偏差（ボラ）: {sigma:.4f}")
````

```python
# ============================

# ざっくり近似： R_yen ≒ R_$ + R_FX
```

ヘッジなし（円換算）

ヘッジあり

実務でのヘッジは、フォワードレート、金利差（キャリー）

私は **S&P500（為替ヘッジあり）** が最も魅力的だと考える。理由は、長期的に高い平均リターンを維持しつつ、為替リスクを取り除くことでリスクが大幅に低減されるためである。ドルの上下に左右されず、米国企業の成長そのものに投資できる点で、安定したリスク・リターン特性を持つと判断した。

# 🔍 さらに別バージョン（参考：学生の個性に応じて）

## ● リスクを取れる学生向け（日本株派）

日経平均はリスクは高いが、その分大きく上昇する局面もある。短期的に値動きを積極的に取りたい投資家には魅力的であり、国内景気や政策の影響を直接受ける点でも身近で理解しやすい。

## ● 為替リスクをあえて取りたい学生向け（円換算S&P500派）

円安が進む局面では、為替ヘッジなしのS&P500は株価上昇と円安の追い風を同時に受け、高いリターンを得られるメリットがある。
